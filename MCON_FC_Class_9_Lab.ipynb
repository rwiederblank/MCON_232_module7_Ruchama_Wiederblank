```python
class Person:
  def __init__(self, name):
    self.name = name

  def speak(self, msg):
    print(msg)

# derived classes
class Student(Person):
  def __init__(self, name, age, courses):
    super().__init__("Tova")
    self.name = name
    self.age = age
    self.courses = courses

  def study(self):
    print("I am studying")

  def walking(self):
    print("I am walking")

  def person_sits(self):
    print("I am sitting")


  #override
  def speak(self, msg):
    print("In student Class", msg)
    super().speak("I am the parent I am speaking")

  def walking(self):
    print("The student is walking")

s1 = Student("Arielle", 19, ["Math", "CS"])
s1.speak("I am speaking")
s1.study()
print(s1.name)
s1.walking()
```
