```python
from abc import ABC, abstractmethod

class Animal(ABC):
  def __init__(self, name:str):
    self.name = name

  def get_name(self):
    return self.name

  @abstractmethod
  def speak(self):
    pass

class Cat(Animal):
  def __init__(self, name):
    self.name = name

  def speak(self):
    return "meow"

class Dog(Animal):
  def speak(self):
    return "woof"


my_animal = Cat("Harvey")
my_dog = Dog("Balter")
```

```python
# word game abstract class
from abc import ABC, abstractmethod

class WordGameController(ABC):
  @abstractmethod
  def play(self):
    pass

  @abstractmethod
  def make_guess(self, letter : str):
    pass

  @abstractmethod
  def check_game_over(self)->bool:
    pass



class TinyWordGame(WordGameController):
  def __init__(self, secret_word:str):
    self.secret_word = secret_word
    self.guesses = set()

  def play(self):
    print("Welcome to the Tiny Word Game, guess the Secret Letter")

  def make_guess(self, letter : str):
    self.guesses.add(letter)
    return letter in self.secret_word

  def check_game_over(self):
    return all (letter in self.guesses for letter in self.secret_word)

game = TinyWordGame("flower")
game.play()
game.make_guess("c")

print(game.check_game_over())
```

```python
class HintWordGame(WordGameController):
  def __init__(self, secret_word:str, hint: str):
    self.secret_word = secret_word
    self.guesses = set()
    self.hint = hint

  def play(self):
    print("Welcome to the Tiny Word Game, guess the Secret Letter")
    print(f"Your secret hint is: {self.hint}")

  def make_guess(self, letter : str):
    self.guesses.add(letter)
    return letter in self.secret_word

  def check_game_over(self):
    return all (letter in self.guesses for letter in self.secret_word)
```

```python
class LimitedGuessWordGame(WordGameController):
  def __init__(self, secret_word:str, guess_limit):
    self.secret_word = secret_word
    self.guesses = set()
    self.guess_limit = guess_limit
    self.num_guesses = 0

  def play(self):
    print("Welcome to the Tiny Word Game, guess the Secret Letter")
    print(f"you only have {self.guess_limit} guesses!")

  def make_guess(self, letter : str):
    self.num_guesses += 1
    self.guesses.add(letter)
    return letter in self.secret_word

  def check_game_over(self):
    if self.num_guesses == self.guess_limit:
      print("Game over! Guesses used up!")
      return all (letter in self.guesses for letter in self.secret_word)
    print(f"You have {self.guess_limit - self.num_guesses} guesses left")
    return all (letter in self.guesses for letter in self.secret_word)

game = LimitedGuessWordGame("flower", 5)
game.play()
game.make_guess("c")

print(game.check_game_over())
```

```python
class Flyer:
  def __init__(self, animal_name):
    self.animal_name = animal_name

  @staticmethod
  def fly():
    print("I am flying")

class Swimmer:
  def __init__(self, swimmer_name):
    self.swimmer_name = swimmer_name

  @staticmethod
  def swim():
    print("I am swimming")


class FlySwimmer(Flyer, Swimmer):
  def __init__(self, flyer_name, swimmer_name):
    Flyer.__init__(self, flyer_name)
    Swimmer.__init__(self, swimmer_name)

duck = FlySwimmer("daffy", "swimming")
duck.swim()
duck.fly()
```

```python
class Vehicle(ABC):

    def __init__(self, brand, speed):
        self._brand = brand
        self._speed = speed

    @abstractmethod
    def get_gas(self):
        pass

# Flying class
class Flying:

    def __init__(self, altitude, airport_destination):
        self._altitude = altitude
        self._airport_destination = airport_destination

# FlyingCar inherits from BOTH classes
class FlyingCar(Vehicle, Flying):

    def __init__(self, brand, speed, altitude, airport_destination):

        Vehicle.__init__(self, brand, speed)
        Flying.__init__(self, altitude, airport_destination)

    def get_gas(self):
        print("got gas ")



# Create object
car1 = FlyingCar("SkyTesla", 180, 5000, "JFK")
print(car1._brand)
print(car1._speed)
print(car1._altitude)
print(car1._airport_destination)
car1.get_gas()
```
